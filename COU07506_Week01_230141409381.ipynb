```python

import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-5, 11)

delta = (n == 0).astype(float)
step = (n >= 0).astype(float)
ramp = n * step
expo = 0.8**n * step

fig, axes = plt.subplots(1, 4, figsize=(14, 3))

signals = [delta, step, ramp, expo]
titles = ["Unit Impulse", "Unit Step",
          "Unit Ramp", "Exponential"]

for ax, x, title in zip(axes, signals, titles):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")
    ax.grid(True)

plt.tight_layout()
plt.show()
```

```python

# TASK 1.2 (A) — Exponential Signals

n = np.arange(0, 21)
a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(13, 3.5))

for ax, a in zip(axes, a_values):
    x = a**n
    ax.stem(n, x)
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n (samples)")
    ax.set_ylabel("Amplitude")
    ax.grid(True)

plt.suptitle("Exponential Signals")
plt.tight_layout()
plt.show()
```

```python

# TASK 1.2 (B) — Discrete-Time Cosine Signals

n = np.arange(0, 41)

x1 = np.cos(2 * np.pi * n / 8)
x2 = np.cos(2 * np.pi * n / 7.5)

plt.figure(figsize=(11, 4))

plt.plot(n, x1, "o-", label="cos(2πn/8)")
plt.plot(n, x2, "s--", label="cos(2πn/7.5)")

plt.xlabel("n (samples)")
plt.ylabel("Amplitude")
plt.title("Comparison of Discrete-Time Cosine Signals")
plt.legend()
plt.grid(True)
plt.show()
```

## Task 1.3 — Working with Audio

```python

# TASK 1.3 — Working with Audio

import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from IPython.display import Audio, display
import os

fname = "hi_there_male_voice.wav"

if os.path.exists(fname):
    fs, x = wavfile.read(fname)

    # Convert stereo to mono
    if x.ndim > 1:
        x = x.mean(axis=1)

    # Convert to floating-point numbers
    x = x.astype(float)

    # Normalize audio amplitude
    peak = np.max(np.abs(x))
    if peak > 0:
        x = x / peak

    # Calculate time
    t = np.arange(len(x)) / fs

    # Display audio information
    print("Sampling Rate:", fs, "Hz")
    print("Duration:", round(len(x)/fs, 2), "seconds")
    print("Total Samples:", len(x))

    # Plot audio waveform
    plt.figure(figsize=(11, 4))
    plt.plot(t, x)
    plt.title("Task 1.3 - Voice Recording")
    plt.xlabel("Time (seconds)")
    plt.ylabel("Normalized Amplitude")
    plt.grid(True)
    plt.show()

    # Play the audio
    display(Audio(x, rate=fs))

else:
    print("ERROR: Audio file not found!")
```

### Analysis Question 1

The sampling rate of the audio recording is **22,050 Hz**, meaning 22,050 samples are taken every second.

For a one-minute mono recording:

Number of samples = Sampling rate × Time

= 22,050 × 60

= **1,323,000 samples**

Therefore, 1,323,000 numbers are required to represent one minute of this audio in mono.

### B. Analysis Questions

**Question 1: What is the sampling rate of your recording? How many numbers are needed to store one minute of it in mono?**

**Answer:**

The sampling rate of the audio file used in Task 1.3 is **22,050 Hz**, meaning 22,050 samples are recorded every second.

Given:

- Sampling rate = 22,050 Hz
- Duration = 60 seconds
- Channels = 1 (mono)

Number of samples = Sampling rate × Duration

= 22,050 × 60

= **1,323,000 samples**

Therefore, 1,323,000 numbers are required to represent one minute of this audio in mono.

**Question 2: Why is a stem plot more appropriate than a continuous line for x(n)?**

**Answer:**

A stem plot is more appropriate because a discrete-time signal x(n) is defined only at integer sample positions. A stem plot displays each individual sample clearly, while a continuous line may incorrectly suggest that signal values exist between the samples.

**Question 3: For a = 1.1 the exponential grows. Which property of a system would such a signal violate if it were an impulse response?**

**Answer:**

If the exponential signal (1.1)^n u(n) were the impulse response of an LTI system, it would violate the **Bounded-Input Bounded-Output (BIBO) stability** condition.

This is because the impulse response increases continuously instead of decreasing, so its absolute sum is infinite. Therefore, the system would be unstable.
